```python
# Calculate skewness and kurtosis

skewness = scaled_data.skew()
kurtosis = scaled_data.kurtosis()

# Combine into a summary DataFrame
stats_data = pd.DataFrame({
    'Skewness': skewness,
    'Kurtosis': kurtosis
})

print(stats_data.to_string(float_format='%.2f'))
```

```python


skew_before = num_clean_wine_data.skew()
kurt_before = num_clean_wine_data.kurtosis()

# Use scaled_data as the transformed_data for plotting
transformed_data = scaled_data

# Calculate skewness and kurtosis AFTER scaling
skew_after = transformed_data.skew()
kurt_after = transformed_data.kurtosis()

def plot_skewness_comparison(col):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

    # Before transformation
    sns.histplot(num_clean_wine_data[col], kde=True, ax=ax1, color='skyblue')
    ax1.set_title(f'Before: Skew={skew_before[col]:.2f}, Kurt={kurt_before[col]:.2f}')

    # After transformation
    sns.histplot(transformed_data[col], kde=True, ax=ax2, color='salmon')
    ax2.set_title(f'After: Skew={skew_after[col]:.2f}, Kurt={kurt_after[col]:.2f}')

    plt.suptitle(f'Distribution of {col}', y=1.05)
    plt.tight_layout()
    plt.show()

# Plot for highly skewed features (e.g., skewness > |0.5|)
for col in num_columns[abs(skew_before) > 0.5]:
    plot_skewness_comparison(col)
```

```python
# Fit a simple linear regression (alcohol → quality) using NumPy and interpret the coefficients.

X = wine_data['alcohol'].values
y = wine_data['quality'].values

X = np.vstack([np.ones(len(X)), X]).T

# X.T is the transpose of the matrix X
coefficients = np.linalg.inv(X.T @ X) @ (X.T @ y)

intercept = coefficients[0]
slope = coefficients[1]

print(f"Intercept (b0): {intercept:.4f}")
print(f"Slope (b1) for 'alcohol': {slope:.4f}")
```
